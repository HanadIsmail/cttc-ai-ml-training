<h1 align='center', style= color:blue> Natural Language Processing (NLP) </h1>

**Definition:** NLP is the field of study concerned with understanding human language and performing meaningful actions based on it.

**Modes of Communication:**
- Speech
- Text

**Core Processes in NLP:**
| Process | Full Form | Description |
|---|---|---|
| **NLU** | Natural Language Understanding | Enables machines to understand and interpret human language |
| **NLG** | Natural Language Generation | Enables machines to generate human-like language as output |

---

## Basic Terminology in NLP

| Term | Definition |
|---|---|
| **Corpus** | A large and structured set of text used for linguistic research or NLP |
| **Document** | A single sentence or a single piece of text |
| **Word** | A combination of characters |
| **Vocabulary** | A collection of unique words |
| **Token** | A piece of a sentence or word, obtained after splitting text |
| **Vector** | A numerical representation that defines the importance of a word |
| **Stop Words** | Commonly used words that carry little meaning (e.g. is, are, this, and) |

---

## Text Preprocessing Pipeline in NLP

1. Tokenization  
2. Stemming  
3. Lemmatization  
4. POS Tagging  
5. Named Entity Recognition  
6. Chunking  

---

### 1. Tokenization
Tokenization is the process of breaking text into smaller units.

There are two types:
- **Sentence Tokenization** — splitting text into individual sentences
- **Word Tokenization** — splitting text into individual words

### 2. Stemming
Stemming is the process of normalizing words into their base or root form by removing suffixes.

### 3. Lemmatization
Lemmatization is the process of converting words into their meaningful dictionary form, known as a lemma.  
It is similar to stemming, but the output of lemmatization is always a proper, valid word.

### 4. POS Tagging
POS stands for **Parts of Speech**. It involves assigning grammatical categories (such as noun, verb, adjective, etc.) to each word in a sentence.

### 5. Named Entity Recognition (NER)
NER is the process of identifying and classifying named entities in text, such as names of people, places, organizations, and dates.

### 6. Chunking
Chunking is the process of grouping words together into meaningful phrases (such as noun phrases) based on their POS tags.

## Stemming

```python
from nltk.stem import PorterStemmer
ps = PorterStemmer()
```

```python
ps.stem('CARS')
```

```python
ps.stem('happines')
```

```python
ps.stem('went')
```

```python
ps.stem('studied')
```

```python
## Lemmatization: it finds the nearest rootword by meaning.
```

```python
from nltk.stem import WordNetLemmatizer
lm = WordNetLemmatizer()
```

```python
lm.lemmatize('went', 'v')
```

```python
lm.lemmatize('eaten', 'v')
```

```python
lm.lemmatize('studied', 'v')
```
