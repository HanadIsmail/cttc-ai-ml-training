```python
fig, axes = plt.subplots(1, 4, figsize=(16, 4))
for ax, col in zip(axes, df.columns):
    sns.boxplot(y=df[col], ax=ax)
    ax.set_title(col)
plt.tight_layout()
plt.show()
```

```python
df = df.melt(var_name='feature', value_name='value')
sns.displot(
    data=df,
    x='value',
    col='feature',
    kde=True,
    height=4,
    aspect=1,
    facet_kws={'sharex': False, 'sharey': False}
)
plt.show()   
```
