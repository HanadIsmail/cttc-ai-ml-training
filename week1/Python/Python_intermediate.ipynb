### Jump Statement and Function 

```python
# break is keyword which is used to bring the program control out of the loop

for i in 'statement':
    if i =='m':
        break
    print(i , end =' ')
    
```

```python
### Continue keyword : 
## * return control of the iteration to the begining of the next iteration of for loop or while Loop
## * all the remaining lines in the preveling iteration of the loop are skipped by the continue keyword

for i in 'statement':
    if i =='e':
        continue
    print(i , end =' ')
```

### Function :

            def fucntion_name(parameters):     -->  // function declaration

```python
### function without arguments

def greet():
    print('Hello')

greet()
```

```python
## function with one argument
def greet1(name):
    print('Hello',name)

greet1('Ali')
```

```python
## function with two argument

def add(n1,n2):
    print(n1+n2)

add(4,5)
```

```python
## function with return value
def add1(n1,n2):
    return n1+n2

add1(10,43)
```

```python
# default argument
def add(m=10, n=20):
    return m+n
add(10,1)
```
