
<pre><h2>Matplotlib</h2> Malplotlib is python library hich is used to data vizualization, it represents the data in graphical format </pre> 
   

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
np.random.randint(1,50,(20))
```

```python
x = np.random.randint(1,100,(200))
y = np.random.randint(1,100,(200))
```

# Line Plot

```python
plt.plot(x,y,color='m')
plt.title('Line Plot',color='g')
plt.xlabel('X-Axis',color='y')
plt.ylabel('Y-Axis',color='r')
plt.show()
```

```python
plt.plot(x,y,'g-')
plt.plot(x,y,'*m')
plt.title('Line Plot',color='g')
plt.xlabel('X-Axis',color='y')
plt.ylabel('Y-Axis',color='r')
plt.xticks(rotation=45)
plt.show()
```

# Bar plot

```python
## bar plot

x = np.random.randint(1,109,(40))
y = np.random.randint(1,100,(40))

plt.bar(x,y,color=['c', 'b', 'm','y','g'],width=1.5)
plt.title('Bar plot',color='b')
plt.annotate('2027',xy=(97, 99),color='r')
plt.show()
```

```python
x = [2,4,6,8]
y = [1,3,5,7]

plt.bar(x,y,color=['c', 'b', 'm','y','g'],width=0.8,label='Boys')
plt.title('Bar plot',color='b')
plt.annotate('2027',xy=(3.7, 3.2),color='r')
a = [1,3,5,7]
b = [3,4,5,3]
plt.bar(a,b,label='girls',color='c')
plt.legend()
plt.show()
```
