```python
import pandas as pd
```

```python
## Creating  a Series
# Convert list into Series

s = pd.Series(['a','b','c','d'])
s
```

```python
type(s)
```

```python
s[2]
```

```python
# Slicing 
s = pd.Series(['a','b','c','d'],index=[10,20,30,40])
s
```

```python
# in slicing we can only use orignal index not the updated index
s[1:4:1]
```

```python
# Tuple to Seies
t = pd.Series((100,200,300,400))
t
```

```python
# Set to series is not possible
```

```python
# dictionary to Series
d = pd.Series({'a':1, 'b':2 , 'c':3, 'd':4, 'e':5})
d
# it treats the key as index value
```

```python
d1 = pd.Series({'a':1, 'b':2 , 'c':3, 'd':4, 'e':5},index=['e','f','g','c','a'])
d1
```
